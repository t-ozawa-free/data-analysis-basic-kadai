```python
# ライブラリのインポート
import matplotlib.pyplot as plt
import pandas as pd
import japanize_matplotlib

# csvファイルの読み込み
try:
    df = pd.read_csv("sample_pandas_6.csv", encoding="utf-8")
    category_df = pd.read_csv("category.csv", encoding="utf-8")
except FileNotFoundError:
    print("ファイルが見つかりません")

# データフレームのマージ
df = pd.merge(df, category_df[["商品番号", "カテゴリー"]], how="inner", on="商品番号")

# カテゴリー内の要素ごとにカウント
category_unique_count = df["カテゴリー"].value_counts()
# 要素ごとの数を棒グラフ化
category_unique_count.plot(kind="bar", title="カテゴリー別の数", rot=45)
# グラフの表示
plt.show()

# 商品番号列ごとに注文数の基本統計量を求める
order_count = df.groupby("商品番号")["注文数"].describe()
print(f"商品番号ごとの注文数の基本統計量は以下の通り\n{order_count}")
```
