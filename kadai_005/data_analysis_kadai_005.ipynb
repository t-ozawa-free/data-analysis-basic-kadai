```python
# 各ライブラリをインポートする
import numpy as np
from matplotlib import pyplot as plt

# 乱数のシードを固定
np.random.seed(3)


# 6個のグラフを表示するための準備（2x3）。
fig, axes = plt.subplots(2, 3, tight_layout=True)

# 折れ線グラフの表示
y0 = [1, 2, -5, 2]
axes[0, 0].plot(y0)

# sin関数を表示
x1 = np.linspace(0, 10, 100)
y1 = 2 + 2 * np.sin(2 * x1)
axes[0, 1].plot(x1, y1, linewidth=2.0)
axes[0, 1].set(xlim=(0, 10), xticks=np.arange(0, 10), ylim=(-1, 5), yticks=np.arange(-1, 6))

# ヒストグラムを表示
x2 = np.random.normal(15, 5, 2000)
axes[0, 2].set(xlim=(-10, 40))
axes[0, 2].hist(x2, bins=30)

# 散布図を表示
x3 = 4 + np.random.normal(0, 2, 60)
y3 = 4 + np.random.normal(0, 2, len(x3))
axes[1, 0].scatter(x3, y3)


plt.show()
```
